- Want to create an animation illustrating the idea of vector addititon for velocities
- One velocity going horizontally ("wind east") and one velocity going vertically ("plane's motion north")

```python
%%manim -qh -v WARNING Test
class Test(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        # Create a plane and show its motion, wind, and net velocity
        # Make all the lines and labels in black color, except the net velocity which should be in red
        plane = SVGMobject(file_name="plane.svg").shift(DOWN*2)
        motion_line = Line(plane.get_center(), plane.get_center() + UP*4).add_tip().set_color(BLACK)
        motion_label = Text("Motion",font="Plus Jakarta Sans").next_to(motion_line, UP).set_color(BLACK)
        wind_line = Line(plane.get_center(), plane.get_center() + RIGHT*2).add_tip().set_color(BLACK)
        wind_label = Text("Wind",font="Plus Jakarta Sans").next_to(wind_line, RIGHT).set_color(BLACK)
        net_velocity = Line(plane.get_center(), plane.get_center() + UP*4 + RIGHT*2).add_tip().set_color(RED)
        net_label = Text("Net Velocity",font="Plus Jakarta Sans").next_to(net_velocity, UP).set_color(RED)
        # Show the motion of the plane and the effect of the wind on its trajectory. Motion without wind first, then add the wind and show the net velocity.
        self.play(FadeIn(plane))
        self.play(Create(motion_line),Write(motion_label))
        self.add(plane) # Readd the plane to the scene to ensure it's on top of the motion line
        self.play(MoveAlongPath(plane, motion_line), run_time=2)
        # Reposition the plane back to the start
        self.play(plane.animate.move_to(DOWN*2))
        # Move the motion lable to avoid overlap with the wind label
        self.play(motion_label.animate.next_to(motion_line, LEFT*1.5))
        self.play(Create(wind_line),Write(wind_label))
        self.play(Create(net_velocity),Write(net_label))
        self.add(plane) # Readd the plane to the scene to ensure it's on top of the motion line
        self.play(MoveAlongPath(plane, net_velocity), run_time=2)
        self.wait(2)
        self.play(FadeOut(plane), FadeOut(motion_line), FadeOut(motion_label), FadeOut(wind_line), FadeOut(wind_label), FadeOut(net_velocity), FadeOut(net_label))
```

```python
%%manim -qh -v WARNING PythagoreanTheoremForward

# Use Pythagorean Theorem to show that the net velocity of the plane is the square root of the sum of the squares of the motion and wind velocities. Show the right triangle formed by the motion, wind, and net velocity vectors, and label the sides with their respective lengths. Use a color scheme to differentiate between the sides of the triangle, with the motion vector in blue, the wind vector in green, and the net velocity vector in red.
# Plane's speed is 12 units, wind speed is 5 units, so net velocity should be 13 units (5-12-13 triangle scaled by 4).

# Adjust positioning to keep everything visible and not overlapping. The motion vector should be vertical, the wind vector should be horizontal, and the net velocity vector should be diagonal. The labels should be positioned next to their respective vectors without overlapping each other or the vectors.
class PythagoreanTheoremForward(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        # Create an origin point for the triangle toward the bottom of the screen to allow space for labels
        origin = ORIGIN + DOWN*2+LEFT*2
        # Create the right triangle
        motion_line = Line(origin, origin + UP*4).set_color(BLUE)
        wind_line = Line(origin, origin + RIGHT*2).set_color(GREEN)
        net_velocity = Line(origin, origin + UP*4 + RIGHT*2).set_color(RED)
        # Create labels for the sides of the triangle
        motion_label = Text("Motion (120 km/h)", font="Plus Jakarta Sans").next_to(motion_line, RIGHT).set_color(BLUE).scale(0.8)
        wind_label = Text("Wind (50 km/h)", font="Plus Jakarta Sans").next_to(wind_line, RIGHT).set_color(GREEN).scale(0.8)

        # calculate the angle of the net velocity line to position the label correctly
        angle = np.arctan(4/2) # arctan(opposite/adjacent) = arctan(4/2)
        net_label = Text("13", font="Plus Jakarta Sans").scale(0.8).move_to(net_velocity.get_center()+LEFT).rotate(angle).set_color(RED) # reduce buffer space to keep it closer to the line
        # Show the triangle and labels
        self.play(Create(motion_line), Write(motion_label))
        self.play(Create(wind_line), Write(wind_label))
        # Animate the "motion" line moving to the right, forming a right triangle with the "wind" line, then transform labels to 5-12 and animate addition of net_velocity line with a value of 13  as the hypotenuse of the triangle.
        self.play(motion_line.animate.shift(RIGHT*2),motion_label.animate.shift(RIGHT*2), run_time=2)
        self.play(transform_labels := Transform(motion_label, Text("12", font="Plus Jakarta Sans").next_to(motion_line, RIGHT).set_color(BLUE).scale(0.8)),
                  Transform(wind_label, Text("5", font="Plus Jakarta Sans").next_to(wind_line, DOWN).set_color(GREEN).scale(0.8)))

        self.play(Create(net_velocity))
        self.play(Write(net_label),run_time=1)
        self.wait(2)
        # Transform labels back to original text to show the relationship between the sides of the triangle and their real-world meanings
        self.play(Transform(motion_label, Text("Motion (120 km/h)", font="Plus Jakarta Sans").move_to(motion_line.get_center() + RIGHT*1.5).set_color(BLUE).scale(0.5)),
                  Transform(wind_label, Text("Wind (50 km/h)", font="Plus Jakarta Sans").next_to(wind_line, DOWN).set_color(GREEN).scale(0.5)),
                  Transform(net_label, Text("Net Velocity (130 km/h)", font="Plus Jakarta Sans").move_to(net_velocity.get_center()+LEFT*0.75).rotate(angle).set_color(RED).scale(0.5))) 

        self.wait(2)

        # Step 1: Fade out the triangle and all text except for the net_label
        self.play(
            FadeOut(motion_line), FadeOut(wind_line), FadeOut(net_velocity),
            FadeOut(motion_label), FadeOut(wind_label),
        )

        # Step 2: Transform "Net Velocity (130 km/h)" into "Speed = 130 km/h" centered as the focal point
        speed_text = Text("Speed = 130 km/h", font="Plus Jakarta Sans").set_color(RED).scale(1.1).move_to(UP*2.5)
        self.play(Transform(net_label, speed_text), run_time=1.5)
        self.wait(1)

        # Step 3: Step through the calculations
        # "Speed is distance over time, so time is distance over speed."
        eq1 = MathTex(r"\text{time}", r"=", r"\frac{\text{distance}}{\text{speed}}").set_color(BLACK).scale(1.3).move_to(DOWN*0.3)
        self.play(Write(eq1), run_time=1.5)
        self.wait(1.5)

        # "The total distance is 2600 km, speed is 130 km per hour"
        eq2 = MathTex(r"\text{time}", r"=", r"\frac{2600 \text{ km}}{130 \text{ km/h}}").set_color(BLACK).scale(1.3).move_to(DOWN*0.3)
        self.play(TransformMatchingTex(eq1, eq2), run_time=1.5)
        self.wait(1.5)

        # "2600 over 130 is the same as 130 times twenty divided by 130"
        eq3 = MathTex(r"\text{time}", r"=", r"\frac{130 \times 20 \text{ km}}{130 \text{ km/h}}").set_color(BLACK).scale(1.3).move_to(DOWN*0.3)
        self.play(TransformMatchingTex(eq2, eq3), run_time=1.5)
        self.wait(1.5)

        # "so divide that 130 out" — cross out the 130s
        numerator_130 = eq3[2][0:3]
        denominator_130 = eq3[2][7:10]
        cancel_num = Line(numerator_130.get_corner(DL), numerator_130.get_corner(UR)).set_color(RED).set_stroke(width=4)
        cancel_den = Line(denominator_130.get_corner(DL), denominator_130.get_corner(UR)).set_color(RED).set_stroke(width=4)
        self.play(Create(cancel_num), Create(cancel_den), run_time=1)
        self.wait(1)

        # "and we get the answer, 20 hours."
        eq4 = MathTex(r"\text{time}", r"=", r"20 \text{ hours}").set_color(BLACK).scale(1.3).move_to(DOWN*0.3)
        self.play(
            FadeOut(cancel_num), FadeOut(cancel_den),
            TransformMatchingTex(eq3, eq4),
            run_time=1.5,
        )
        self.wait(2)
```
